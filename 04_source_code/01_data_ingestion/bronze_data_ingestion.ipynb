# End-to-End E-Commerce Shipping Analytics
## Bronze Layer Data Ingestion

This notebook ingests the source CSV file from OneLake Files into the Bronze layer of the Lakehouse using Spark SQL. 
The raw source data is preserved as a managed Delta table for downstream transformation and analytics.

```sql
%%sql

-- ============================================================
-- Project: E-Commerce Shipping Analytics
-- Purpose: Ingest the source CSV file into the Bronze layer
-- Input:   Files/E_Commerce_Shipping_Data_Train.csv
-- Output:  bronze_shipping_data
-- ============================================================


-- Step 1: Remove the temporary view if it already exists
DROP VIEW IF EXISTS v_raw_csv_import;


-- Step 2: Create a temporary view over the source CSV file
-- The view promotes the CSV header and infers the source schema
CREATE OR REPLACE TEMPORARY VIEW v_raw_csv_import
USING csv
OPTIONS (
    path 'Files/E_Commerce_Shipping_Data_Train.csv',
    header 'true',
    inferSchema 'true'
);


-- Step 3: Materialize the temporary view as a managed Delta table
CREATE OR REPLACE TABLE bronze_shipping_data
USING DELTA
AS
SELECT *
FROM v_raw_csv_import;


-- Step 5: Validate the number of records loaded
SELECT
    COUNT(*) AS total_records
FROM bronze_shipping_data;


-- Step 6: Preview the Bronze data
SELECT *
FROM bronze_shipping_data
LIMIT 10;
```

## Notebook Summary

The source CSV file was ingested into the Bronze layer and stored as the `bronze_shipping_data` Delta table.
